```python
name="khalid"
age=22.87328743827
print("name:"+name+",Age:"+str(age))
```

```python
print("name:%s ,Age:%d"%(name,age))
```

```python
print("name:{0},age:{1:.4f}".format(name,age))
```

```python
print(f"name:{name:<10},age:{age:0.2f}")
print(f"name:{name:<10},age:{age:0.2f}")
print(f"name:{name:<10},age:{age:0.2f}")
```

```python
lstname=[("hamzah",77),("abdalmohaymen",82),("adel",91),("ali",66)]
for name,age in lstname:
    print(f"name:{name:<15},age:{age:0.2f}")
```
