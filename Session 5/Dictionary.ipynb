```python
d1={"name":"ahmad","salary":120,"city":"aleppo"}
d2={"name":"haider","salary":220,"city":"Idlib"}
d3={"name":"hamza","salary":170,"city":"aleppo"}
d4={"name":"mostfa","salary":180,"city":"Idlib"}
d5={"name":"mohamma","salary":250,"city":"Idlib"}
lst=[d1,d2,d3,d4,d5]
#list of dictionares
```

```python
print(len(lst))
```

```python
lst
```

```python
for emp in lst:
    print(emp["name"],emp["salary"],sep=",")
```

```python
total=0
for emp in lst:
    if emp["city"]=="aleppo":
        total+=emp["salary"]
print(total)
    
```

```python
total=0
for emp in lst:
    if emp["city"].lower()=="Idlib".lower():
        total+=emp["salary"]
print(total)
    
```

```python
d1={"name":"ahmad","salary":-120,"city":"aleppo"}
d2={"name":"haider","salary":-220,"city":"Idlib"}
d3={"name":"hamza","salary":70,"city":"aleppo"}
d4={"name":"mostfa","salary":-180,"city":"Idlib"}
d5={"name":"mohamma","salary":-50,"city":"Idlib"}
lst=[d1,d2,d3,d4,d5]
#list of dictionares
```

```python
maxSalary=lst[0]["salary"]
for emp in lst:
    if emp["salary"]>maxSalary:
        maxSalary=emp["salary"]
print(maxSalary)
    
```
