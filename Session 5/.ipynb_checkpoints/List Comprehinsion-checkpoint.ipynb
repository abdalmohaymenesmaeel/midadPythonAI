```python
import time
start= time.time()
lst=[]
for i in range(1,1_000_000):
    if i%2==0:
        lst.append(i**2)
end= time.time()
print("Using for loop:",end-start,"seconds")
```

```python
start=time.time()
lst=[i*i for i in range(1,1_000_000)]
end=time.time()
print("Using for comprehension:",end-start,"seconds")
```

```python
celsius=[1,10,20]
fahr=[]
for i in celsius:
    fahr.append((i*9/5)+32)
fahr
```

```python
start=time.time()
m=[]
for i in range(3000):
    innerlist=[]
    for j in range(3000):
        innerlist.append(i*j)
    m.append(innerlist)
end=time.time()
print(end-start)

    
        
```

```python
start=time.time()
m1=[[j*i for j in range(3000)] for i in range(3000)]
end=time.time()
print(end-start)
```
