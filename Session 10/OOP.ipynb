```python
class Employee:
    def __init__(self):
        self.name="ahmad"
        self.lname="khateeb"
        self.salary=270
    def print(self):
        print(f"{self.name} {self.lname} salary is {self.salary}")
```

```python
emp1=Employee()
print(emp1)
```

```python
emp1.print()
```

```python
class EmployeeV1:
    def __init__(self,name,lname,salary):
        self.name=name
        self.lname=lname
        self.salary=salary
    def print(self):
        print(f"{self.name} {self.lname} salary is {self.salary}")
```

```python
emp=EmployeeV1("abdalmohaymen","alesmaeel",400)
```

```python
emp.print()
```
