```python
#changing State
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def printData(self):
        print(f"name:{self.name},age:{self.age}")
    def changeName(self,newName):
        if len(newName) > 1:
            self.name=newName
    def changeAge(self,newAge):
        if newAge>=0:
            self.age=newAge
```

```python
std=Student("ahmad",33)
std.changeName("")
std.changeAge(-3)
std.printData()
```

```python
std.name=""
std.printData()
```

```python
#proprty (getter and setter)
class Student:
    def __init__(self,name,age):
        self.__name=name
        self.age=age
    def printData(self):
        print(f"name:{self.__name},age:{self.age}")
    def name(self):
        return self.__name
    def name(self,newName):
        if len(newName) > 1:
            self.__name=newName
std=Student("ahmad",33)
std.name="ghfhgf"
std.printData()
```

```python
#proprty (getter and setter)
class Student:
    def __init__(self,name,age):
        self.__name=name
        self.age=age
    def printData(self):
        print(f"name:{self.__name},age:{self.age}")
    @property
    def name(self):
        return self.__name
    @name.setter
    def name(self,newName):
        if len(newName) > 1:
            self.__name=newName
std=Student("ahmad",33)
std.name="ghfhgf"
std.printData()
std.name=""
std.printData()
```

```python
import math
class Circle:
    def __init__(self,radius):
        self.counter=0
        self.counter+=1
        self.radius=radius
    def getArea(self):
        return math.pi*self.radius**2
    def getPrim(self):
        return math.pi*self.radius*2
```

```python
c=Circle(10)
c.getArea()
```

```python
c2=Circle(55)
c3=Circle(15)
c4=Circle(5)
```

```python
c4.counter
```
