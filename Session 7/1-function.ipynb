```python
def getSum(lst):
    total=0
    for i in lst:
        total+=i
    return total
def getAverage(lst):
    return getSum(lst)/len(lst)

import time
start=time.time()
lst = list(range(1,100001))
avg=getAverage(lst)
print(avg)
end=time.time()
print(end-start)
```

```python
def fun2(n1,n2,state=False):
    if state==True:
        return n1+n2
    else:
        return n1*n2
fun2(10,20,True)
#default parametr
#optional parameter
```

```python
#variable length *args
def getSum(n1,n2):
    return n1+n2
def getSum(n1,n2,n3):
    return n1+n2+n3
def getSum(*args):
    total=0
    for i in args:
        total+=i
    return total
print(getSum(10,20))
print(getSum(10,20,30))
print(getSum(10,20,30,40))
print(getSum(10,20,30,40,50))
```
